### 2.7 Balance each class in Training DATASET

```python
df_train['duplicates'] = 'unduplicated'
```

```python
def duplicate(df: pd.DataFrame, target_class: list[str], max_per_class: int):
  augmented_df = df.copy()

  for i in tqdm(target_class, desc='BALANCING THE CLASS', unit = 'image'):

    class_subset = df[(df['cell_type_idx'] == i) & (df['duplicates'] == 'unduplicated')]
    if class_subset.empty:
                print(f"Warning: No unduplicated rows for class {i}. Skipping.")
                continue

    current_count = (augmented_df['cell_type_idx'] == i).sum()
    new_row = []

    while current_count< max_per_class:
      random_row = class_subset.sample(n=1).iloc[0]

      row_dict = random_row.to_dict()
      row_dict['duplicates'] = 'duplicated'
      new_row.append(row_dict)
      current_count += 1

    if new_row:
      augmented_df = pd.concat([augmented_df, pd.DataFrame(new_row)], ignore_index=True)

  return augmented_df
```

## Training

### 3.1 Create Dataloader

```python
df_train = df_train.reset_index(drop = True)
df_val = df_val.reset_index(drop = True)
```

```python
import torch
from torch.utils.data import Dataset
from PIL import Image
# Create our class of PyTorch dataloader
class SkinDataLoader(Dataset):
  def __init__(self, df, transforms=None):
    # NOTE: df = pandas Dataframe or list that containing a filepaths and labels
    # transform = torchvision.transforms (or your own func) to apply to each sample
    self.df = df
    self.transform = transforms

  def __len__(self):
    #Return number of samples
    return len(self.df)

  def __getitem__(self, index):
    X = Image.open(self.df["path"][index])
    y = torch.tensor(int(self.df['cell_type_idx'][index]))

    if self.transform:
      X = self.transform(X)
    return X, y
```

### 3.2 Create Transformer

```python
from torch.utils.data import DataLoader
from torchvision import transforms
BATCH_SIZE = 10

train_tfms = transforms.Compose([
    #transforms.Resize((224,224)),
    transforms.Resize((512,512)),
    transforms.RandomRotation(degrees=30),
    transforms.RandomHorizontalFlip(p=0.25),
    transforms.RandomVerticalFlip(p=0.25),
    transforms.ColorJitter(brightness=0.1, contrast=0.1, hue=0.1),
    transforms.ToTensor()
])

training_set = SkinDataLoader(df_train, transforms = train_tfms)
train_dataloader = DataLoader(training_set, batch_size = BATCH_SIZE , shuffle = True, num_workers=4)

validation_set = SkinDataLoader(df_val, transforms = train_tfms)
val_dataloader = DataLoader(validation_set, batch_size= BATCH_SIZE, shuffle = True, num_workers=4)

train_dataloader, val_dataloader
```

### 3.3 Create DenseNet Architecture (DenseNet121)

```python
# feature_extract is a boolean that defines if we are finetuning or feature extracting.
# If feature_extract = False, the model is finetuned and all model parameters are updated.
# If feature_extract = True, only the last layer parameters are updated, the others remain fixed.
def set_parameter_requires_grad(model, feature_extracting):
    if feature_extracting:
        for param in model.parameters():
            param.requires_grad = False
```

```python
import torch
from torch import nn
import torchvision
from torchvision import models

def Densenet121(num_classes, feature_extract, use_pretrained=True):
  # Densenet121
  model_ft = models.densenet121(pretrained=use_pretrained)
  set_parameter_requires_grad(model_ft, feature_extract)
  num_ftrs = model_ft.classifier.in_features
  model_ft.classifier = nn.Linear(num_ftrs, num_classes)
  #input_size = 224
  input_size = 450

  return model_ft, input_size
```

### 3.4 Create instance of denseNet

```python
if torch.cuda.is_available():
  device = 'cuda'
else:
  device = 'cpu'
print(device)
```

```python
model_0, input_size = Densenet121(
                                  num_classes = df_train['dx'].nunique(),
                                  feature_extract = False,
                                  use_pretrained=True
)

model_0.to(device)
```

### 3.5 Define Training and Validation steps

```python
!pip install torchmetrics -q
```

```python
# this function is used during training process, to calculation the loss and accuracy
class AverageMeter(object):
    def __init__(self):
        self.reset()

    def reset(self):
        self.val = 0
        self.avg = 0
        self.sum = 0
        self.count = 0

    def update(self, val, n=1):
        self.val = val
        self.sum += val * n
        self.count += n
        self.avg = self.sum / self.count
```

```python
total_loss_train, total_acc_train = [],[]
def train(train_loader, model, criterion, optimizer, epoch):
    model.train()
    train_loss = AverageMeter()
    train_acc = AverageMeter()
    curr_iter = (epoch - 1) * len(train_loader)
    for i, data in enumerate(train_loader):
        images, labels = data
        N = images.size(0)
        # print('image shape:',images.size(0), 'label shape',labels.size(0))
        images = Variable(images).to(device)
        labels = Variable(labels).to(device)

        optimizer.zero_grad()
        outputs = model(images)

        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        prediction = outputs.max(1, keepdim=True)[1]
        train_acc.update(prediction.eq(labels.view_as(prediction)).sum().item()/N)
        train_loss.update(loss.item())
        curr_iter += 1
        if (i + 1) % 100 == 0:
            print('[epoch %d], [iter %d / %d], [train loss %.5f], [train acc %.5f]' % (
                epoch, i + 1, len(train_loader), train_loss.avg, train_acc.avg))
            total_loss_train.append(train_loss.avg)
            total_acc_train.append(train_acc.avg)
    return train_loss.avg, train_acc.avg
```

```python
from torch.autograd import Variable
def validate(val_loader, model, criterion, optimizer, epoch):

    # Insert this at the top of validate(), before model.eval()
    unique_labels = set()
    for _, labels in val_loader:
        unique_labels.update(labels.tolist())
    print("VAL labels:", sorted(unique_labels))

    model.eval()
    val_loss = AverageMeter()
    val_acc = AverageMeter()
    with torch.no_grad():
        for i, data in enumerate(val_loader):
            images, labels = data
            N = images.size(0)
            images = Variable(images).to(device)
            labels = Variable(labels).to(device)

            outputs = model(images)
            prediction = outputs.max(1, keepdim=True)[1]

            val_acc.update(prediction.eq(labels.view_as(prediction)).sum().item()/N)

            val_loss.update(criterion(outputs, labels).item())

    print('------------------------------------------------------------')
    print('[epoch %d], [val loss %.5f], [val acc %.5f]' % (epoch, val_loss.avg, val_acc.avg))
    print('------------------------------------------------------------')
    return val_loss.avg, val_acc.avg
```

### 3.6 Define Loss Function and Optimizer

```python
from torch import optim

loss_fn = nn.CrossEntropyLoss().to(device)
optimizer = optim.Adam(model_0.parameters(), lr= 1e-3)
```

### 3.7 Training Model

```python
torch.manual_seed(101)

epoch_num = 5
best_val_acc = 0
total_loss_val, total_acc_val = [],[]
for epoch in tqdm(range(1, epoch_num+1)):

    loss_train, acc_train = train(train_dataloader, model_0, loss_fn, optimizer, epoch)
    loss_val, acc_val = validate(val_dataloader, model_0, loss_fn, optimizer, epoch)
    total_loss_val.append(loss_val)
    total_acc_val.append(acc_val)
    if acc_val > best_val_acc:
        best_val_acc = acc_val
        print('*****************************************************')
        print('best record: [epoch %d], [val loss %.5f], [val acc %.5f]' % (epoch, loss_val, acc_val))
        print('*****************************************************')
```
